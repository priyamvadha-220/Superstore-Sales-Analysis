```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Sample - Superstore.csv", encoding='latin1')
```

```python
# 1. Which customers ordered most frequently?

top_customers = (df['Customer Name'].value_counts().head(10).reset_index())

top_customers.columns = ['Customer Name', 'Number of Orders']

top_customers
```

```python
print(df['State'].unique())
```

```python
# 2. Enter a state name and display its total sales and profit.

state = input("Enter State Name: ")

state_data = df[df['State'] == state]

total_sales = state_data['Sales'].sum()
total_profit = state_data['Profit'].sum()

print("\nTotal Sales:", total_sales)
print("Total Profit:", total_profit)
```

```python
# 3. Which orders generated the highest profit?

top_profit_orders = (df[['Order ID', 'Customer Name', 'Sales', 'Profit']].sort_values(by='Profit', ascending=False).head(10))

top_profit_orders
```

```python
print(df['Customer Name'].unique())
```

```python
# 4. Enter a customer name and display all their orders.

customer = input("Enter Customer Name: ")

customer_orders = df[df['Customer Name'] == customer]

customer_orders
```

```python
# 5. What is the standard deviation of sales and profit?

print("Sales Standard Deviation:", np.std(df['Sales']))
print("Profit Standard Deviation:", np.std(df['Profit']))
```

```python
# 6. What are the mean, median, maximum, and minimum sales values?

print("Mean Sales:", np.mean(df['Sales']))
print("Median Sales:", np.median(df['Sales']))
print("Maximum Sales:", np.max(df['Sales']))
print("Minimum Sales:", np.min(df['Sales']))
```

```python
# 7. Which category generated the highest total sales?

category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

print(category_sales)

plt.figure(figsize=(4,3))

sns.barplot(x=category_sales.index, y=category_sales.values)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
# 8. Which sub-category generated the highest profit?

subcategory_profit = df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False)

print(subcategory_profit)

plt.figure(figsize=(7,3))

sns.barplot(
    x=subcategory_profit.index,
    y=subcategory_profit.values, color='cyan')

plt.xticks(rotation=70)

plt.title("Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Profit")

plt.show()
```

```python
# 9. Which state has the highest sales?

state_sales = df.groupby('State')['Sales'].sum()

top_states = state_sales.sort_values(ascending=False).head(10)

plt.figure(figsize=(7,4))

sns.barplot(x=top_states.values, y=top_states.index, color='turquoise')

plt.title("Top 10 States by Total Sales", fontsize=10)
plt.xlabel("Total Sales", fontsize=10)
plt.ylabel("State", fontsize=10)

plt.show()
```

```python
# 10. Which region is the most profitable?

region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)

plt.figure(figsize=(6,3))

sns.barplot(x=region_profit.index,y=region_profit.values,
    hue=region_profit.index,
    palette='viridis')

plt.title("Profit by Region", fontsize=12)
plt.xlabel("Region", fontsize=10)
plt.ylabel("Total Profit", fontsize=10)

plt.show()
```
